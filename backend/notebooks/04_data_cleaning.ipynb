```python
# ============================================
# 📚 Import Libraries
# ============================================

import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
```

```python
# ============================================
# 📥 Load Datasets
# ============================================

results = pd.read_csv("../data/raw/results.csv")

wcmatches = pd.read_csv("../data/raw/wcmatches.csv")

worldcups = pd.read_csv("../data/raw/worldcups.csv")
```

```python
# 🇩🇪 Data Standardization: Germany Unification

# Automatically detect dataframe variables
possible_dfs = ["df", "matches", "results", "world_cup", "team_metrics", "data"]

for df_name in possible_dfs:

    if df_name in globals():

        dataframe = globals()[df_name]

        country_columns = [
            col for col in dataframe.columns
            if any(keyword in col.lower()
                   for keyword in ["team", "country", "winner", "home", "away"])
        ]

        for col in country_columns:
            dataframe[col] = dataframe[col].replace("West Germany", "Germany")

        print(f"✅ Germany standardization applied to: {df_name}")
```

## ✅ Final Dataset Validation

Confirming dataset quality after preprocessing.

```python
# ============================================
# 📊 Initial Dataset Overview
# ============================================

print(f"Results Shape: {results.shape}")

print(f"World Cup Matches Shape: {wcmatches.shape}")

print(f"World Cups Shape: {worldcups.shape}")
```

## ⚙️ Data Standardization

Applying consistency rules across categorical variables.

```python
# ============================================
# 🌍 Team Name Standardization
# ============================================

team_name_mapping = {

    # Germany
    "West Germany": "Germany",
    "East Germany": "Germany",

    # Soviet Union / Russia
    "Soviet Union": "Russia",
    "USSR": "Russia",

    # Czech Republic
    "Czechoslovakia": "Czech Republic",

    # Yugoslavia
    "Yugoslavia": "Serbia",

    # Congo
    "Zaire": "DR Congo",

    # Korea
    "Korea Republic": "South Korea",

    # Iran
    "IR Iran": "Iran",

    # China
    "China PR": "China",

    # USA
    "USA": "United States"
}
```

```python
# ============================================
# 🔄 Apply Team Name Standardization
# ============================================

results_columns = [
    "home_team",
    "away_team"
]

for column in results_columns:

    results[column] = (
        results[column]
        .replace(team_name_mapping)
    )

wcmatches_columns = [
    "home_team",
    "away_team",
    "winning_team",
    "losing_team"
]

for column in wcmatches_columns:

    wcmatches[column] = (
        wcmatches[column]
        .replace(team_name_mapping)
    )

worldcups["winner"] = (
    worldcups["winner"]
    .replace(team_name_mapping)
)
```

```python
# ============================================
# 🚫 Remove Non-FIFA Teams
# ============================================

non_fifa_teams = [

    "Jersey",
    "Guernsey",
    "Isle of Man",
    "Asturias",
    "Surrey",
    "Yorkshire",
    "Biafra",
    "Maule Sur",
    "Padania",
    "Catalonia",
    "Canary Islands",
    "Corsica",
    "Occitania",
    "Saarland",
    "Vietnam Republic",
    "Bohemia",
    "Cascadia"
]
```

```python
# ============================================
# 🧹 Filter Invalid Teams
# ============================================

results = results[
    (~results["home_team"].isin(non_fifa_teams)) &
    (~results["away_team"].isin(non_fifa_teams))
]
```

```python
# ============================================
# 📉 Dataset Shape After Cleaning
# ============================================

print(f"Clean Results Shape: {results.shape}")
```

## 🔍 Missing Values Analysis

Identifying and handling incomplete records.

```python
# ============================================
# 🔍 Missing Values Analysis
# ============================================

results.isnull().sum()
```

```python
# ============================================
# 🧼 Handle Missing Values
# ============================================

results["home_score"] = (
    results["home_score"]
    .fillna(0)
)

results["away_score"] = (
    results["away_score"]
    .fillna(0)
)
```

## 🧪 Data Type Validation

Ensuring proper data formats and structures.

```python
# ============================================
# 📅 Date Formatting
# ============================================

results["date"] = pd.to_datetime(
    results["date"]
)

results["year"] = (
    results["date"]
    .dt.year
)
```
