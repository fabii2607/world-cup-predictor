```python
# 🛠️ Adding 2022 World Cup Data

new_row = pd.DataFrame({
    "year": [2022],
    "host": ["Qatar"],
    "winner": ["Argentina"],
    "second": ["France"],
    "third": ["Croatia"],
    "fourth": ["Morocco"],
    "goals_scored": [172],
    "teams": [32],
    "games": [64],
    "attendance": [3404252]
})

worldcups = pd.concat(
    [worldcups, new_row],
    ignore_index=True
)
```

```python
worldcup_winners = (
    worldcups["winner"]
    .value_counts()
)

worldcup_winners
```

```python
# 🇩🇪 Germany Name Standardization

worldcups["winner"] = worldcups["winner"].replace(
    "West Germany",
    "Germany"
)
```

```python
worldcup_winners = (
    worldcups["winner"]
    .value_counts()
)

worldcup_winners
```

```python
# 📊 FIFA World Cup Titles by Country

plt.figure(figsize=(12,6)
plt.grid(alpha=0.3))

sns.barplot(
    x=worldcup_winners.values,
    y=worldcup_winners.index,
    palette="viridis"
)

plt.title("FIFA World Cup Titles by Country", fontsize=16)
plt.xlabel("Number of Titles", fontsize=12)
plt.ylabel("Country", fontsize=12)

plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.show()

plt.tight_layout()
```

```python
# ⚽ Goals Scored per World Cup

goals_per_cup = (
    worldcups[["year", "goals_scored"]]
)

goals_per_cup.head()
```

```python
# 📈 Goals Scored in Each World Cup

plt.figure(figsize=(14,6)
plt.grid(alpha=0.3))

sns.lineplot(
    data=goals_per_cup,
    x="year",
    y="goals_scored",
    marker="o"
)

plt.title("Goals Scored per FIFA World Cup")
plt.xlabel("Year")
plt.ylabel("Goals Scored")

plt.show()
```

```python
# ⚽ Goals Per Match Calculation

worldcups["goals_per_match"] = (
    worldcups["goals_scored"] /
    worldcups["games"]
)

worldcups[["year", "goals_scored", "games", "goals_per_match"]]
```

```python
# 📈 Average Goals Per Match by World Cup

plt.figure(figsize=(14,6)
plt.grid(alpha=0.3))

sns.lineplot(
    data=worldcups,
    x="year",
    y="goals_per_match",
    marker="o",
    color="green",
    linewidth=2.5
)

plt.title("Average Goals Per Match in FIFA World Cups")
plt.xlabel("Year")
plt.ylabel("Goals Per Match")

plt.grid(alpha=0.3)

plt.show()
```

# 📌 Key Insights


- Earlier FIFA World Cups were significantly more offensive, with very high average goals per match.
- The 1954 World Cup stands out as one of the highest-scoring tournaments in history.
- After the mid-20th century, average goals per match declined considerably and became more stable.
- Modern World Cups typically maintain averages between 2 and 3 goals per match.
- This trend may reflect tactical evolution, stronger defensive systems,
  increased competitiveness, and improved physical preparation in modern football.
- Recent tournaments show slight offensive recovery, suggesting a more dynamic modern playing style.

```python
# 🏟️ Host Country vs World Cup Winner

worldcups[["year", "host", "winner"]]
```

```python
# 🏆 Did the Host Country Win?

worldcups["host_winner"] = (
    worldcups["host"] == worldcups["winner"]
)

worldcups[["year", "host", "winner", "host_winner"]]
```

```python
# 📊 Number of Host Winners

host_wins = worldcups["host_winner"].value_counts()

host_wins
```

```python
# 📈 Host Country World Cup Wins

plt.figure(figsize=(8,5)
plt.grid(alpha=0.3))

sns.barplot(
    x=host_wins.index.astype(str),
    y=host_wins.values,
    hue=host_wins.index.astype(str),
    palette="crest",
    legend=False
)

plt.title("Did Host Countries Win the World Cup?")
plt.xlabel("Host Won Tournament")
plt.ylabel("Count")

plt.grid(axis="y", alpha=0.3)

plt.show()

plt.tight_layout()
```

```python
# 🥇 Host Country Top 4 Analysis

worldcups["host_top4"] = (
    (worldcups["host"] == worldcups["winner"]) |
    (worldcups["host"] == worldcups["second"]) |
    (worldcups["host"] == worldcups["third"]) |
    (worldcups["host"] == worldcups["fourth"])
)

worldcups[["year", "host", "host_top4"]]
```

```python
# 📊 Host Top 4 Count

host_top4_counts = (
    worldcups["host_top4"]
    .value_counts()
)

host_top4_counts
```

```python
# 📈 Host Countries Reaching Top 4

plt.figure(figsize=(8,5)
plt.grid(alpha=0.3))

sns.barplot(
    x=host_top4_counts.index.astype(str),
    y=host_top4_counts.values,
    hue=host_top4_counts.index.astype(str),
    palette="viridis",
    legend=False
)

plt.title("Did Host Countries Reach Top 4?")
plt.xlabel("Host Reached Top 4")
plt.ylabel("Count")

plt.grid(axis="y", alpha=0.3)

plt.show()

plt.tight_layout()
```

# 📌 Key Insights


- Host countries do not frequently win the FIFA World Cup, but they often achieve strong tournament performances.
- Approximately half of the host nations reached the Top 4 positions in World Cup history.
- These findings suggest that hosting the tournament may provide a meaningful competitive advantage.
- Possible factors include crowd support, reduced travel fatigue, climate familiarity,
  and psychological confidence associated with playing at home.
- Host nation influence may become an important feature in future predictive modeling stages.

```python
# 🌍 Country to Continent Mapping

continent_map = {
    "Brazil": "South America",
    "Argentina": "South America",
    "Uruguay": "South America",

    "Germany": "Europe",
    "Italy": "Europe",
    "France": "Europe",
    "England": "Europe",
    "Spain": "Europe"
}
```

```python
# 🌍 Winner Continents

worldcups["winner_continent"] = (
    worldcups["winner"]
    .map(continent_map)
)

worldcups[["year", "winner", "winner_continent"]]
```

```python
# 📊 World Cup Titles by Continent

continent_titles = (
    worldcups["winner_continent"]
    .value_counts()
)

continent_titles
```

```python
# 📈 Continental Dominance

plt.figure(figsize=(10,6)
plt.grid(alpha=0.3))

sns.barplot(
    x=continent_titles.index,
    y=continent_titles.values,
    hue=continent_titles.index,
    palette="Spectral",
    legend=False
)

plt.title("FIFA World Cup Titles by Continent")
plt.xlabel("Continent")
plt.ylabel("Number of Titles")

plt.grid(axis="y", alpha=0.3)

plt.show()

plt.tight_layout()
```

```python
# 📈 Continental Winners Through Time

plt.figure(figsize=(14,6)
plt.grid(alpha=0.3))

sns.scatterplot(
    data=worldcups,
    x="year",
    y="winner_continent",
    hue="winner_continent",
    palette="Spectral",
    s=150
)

plt.title("World Cup Winners by Continent Through Time")
plt.xlabel("Year")
plt.ylabel("Continent")

plt.grid(alpha=0.3)

plt.show()
```

# 📌 Key Insights


- Europe currently holds the highest number of FIFA World Cup titles,
  although the difference compared to South America remains relatively small.
- Early World Cup history showed stronger South American dominance,
  while Europe became increasingly dominant in modern tournaments.
- The competitive balance between Europe and South America highlights
  the historical concentration of elite football nations within these two continents.
- Modern European dominance may be associated with stronger domestic leagues,
  infrastructure investment, tactical evolution, and player development systems.
- Despite fewer titles, South America maintains remarkable historical efficiency
  with relatively few countries consistently competing at the highest level.

## ⚔️ Matchup & Finals Analysis

Identifying recurring rivalries and historical final matchups.

```python
# ⚔️ Creating Matchup Identifier

wcmatches["matchup"] = (
    wcmatches[["home_team", "away_team"]]
    .apply(lambda x: " vs ".join(sorted(x)), axis=1)
)

wcmatches[["home_team", "away_team", "matchup"]].head()
```

```python
# 📊 Most Frequent Matchups

top_matchups = (
    wcmatches["matchup"]
    .value_counts()
    .head(10)
)

top_matchups
```

```python
# 📈 Top 10 World Cup Matchups

plt.figure(figsize=(14,6)
plt.grid(alpha=0.3))

sns.barplot(
    x=top_matchups.values,
    y=top_matchups.index,
    hue=top_matchups.index,
    palette="rocket",
    legend=False
)

plt.title("Top 10 Most Frequent World Cup Matchups")
plt.xlabel("Number of Matches")
plt.ylabel("Matchup")

plt.grid(axis="x", alpha=0.3)

plt.show()

plt.tight_layout()
```

```python
# 🏆 World Cup Finals Only

final_matches = (
    wcmatches[wcmatches["stage"] == "Final"]
)

final_matches[["year", "home_team", "away_team"]]
```

```python
# ⚔️ Most Frequent Final Matchups

final_matches["final_matchup"] = (
    final_matches[["home_team", "away_team"]]
    .apply(lambda x: " vs ".join(sorted(x)), axis=1)
)

top_final_matchups = (
    final_matches["final_matchup"]
    .value_counts()
)

top_final_matchups
```

```python
# 📈 Most Frequent World Cup Finals

plt.figure(figsize=(12,5)
plt.grid(alpha=0.3))

sns.barplot(
    x=top_final_matchups.values,
    y=top_final_matchups.index,
    hue=top_final_matchups.index,
    palette="magma",
    legend=False
)

plt.title("Most Frequent FIFA World Cup Finals")
plt.xlabel("Number of Finals")
plt.ylabel("Final Matchup")

plt.grid(axis="x", alpha=0.3)

plt.show()

plt.tight_layout()
```

```python
# 🇩🇪 Standardizing Germany Names

columns_to_fix = [
    "home_team",
    "away_team",
    "winning_team",
    "losing_team"
]

for col in columns_to_fix:
    wcmatches[col] = wcmatches[col].replace(
        "West Germany",
        "Germany"
    )
```

```python
# ⚔️ Rebuilding Matchups

wcmatches["matchup"] = (
    wcmatches[["home_team", "away_team"]]
    .apply(lambda x: " vs ".join(sorted(x)), axis=1)
)

top_matchups = (
    wcmatches["matchup"]
    .value_counts()
    .head(10)
)

top_matchups
```

```python
# 🏆 Rebuilding Final Matchups

final_matches = (
    wcmatches[wcmatches["stage"] == "Final"]
)

final_matches["final_matchup"] = (
    final_matches[["home_team", "away_team"]]
    .apply(lambda x: " vs ".join(sorted(x)), axis=1)
)

top_final_matchups = (
    final_matches["final_matchup"]
    .value_counts()
)

top_final_matchups
```

# 📌 Key Insights


- Historical inconsistencies were identified between different datasets,
  particularly regarding the use of 'West Germany' and 'Germany'.
- Standardizing historical country names was necessary to ensure accurate matchup aggregation.
- Data consistency across datasets is essential for reliable statistical analysis
  and future predictive modeling tasks.
