```python
# ============================================
# 📚 Import Libraries
# ============================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import random

from sklearn.ensemble import GradientBoostingClassifier
from sklearn.preprocessing import LabelEncoder

import warnings
warnings.filterwarnings("ignore")
```

## 🗂️ Dataset Preparation

Loading datasets and preparing tournament analysis structures.

```python
# ============================================
# 📥 Load Enhanced Dataset
# ============================================

match_df = pd.read_csv(
    "../data/processed/match_dataset_enhanced.csv"
)

match_df.head()
```

```python
# ============================================
# ⚙️ Feature Selection
# ============================================

features = [

    "home_avg_points_last_5",
    "away_avg_points_last_5",

    "home_avg_goals_scored_last_5",
    "away_avg_goals_scored_last_5",

    "home_avg_goals_conceded_last_5",
    "away_avg_goals_conceded_last_5",

    "home_avg_goal_diff_last_5",
    "away_avg_goal_diff_last_5",

    "home_win_streak",
    "away_win_streak",

    "home_clean_sheets_last_5",
    "away_clean_sheets_last_5",

    "home_momentum_score",
    "away_momentum_score",

    "home_elo",
    "away_elo",

    "elo_difference"
]

X = match_df[features].fillna(0)

y = match_df["target"]
```

```python
# ============================================
# 🔤 Encode Target
# ============================================

encoder = LabelEncoder()

y_encoded = encoder.fit_transform(y)
```

```python
# ============================================
# 🚀 Train Tournament Model
# ============================================

model = GradientBoostingClassifier(

    n_estimators=200,

    learning_rate=0.05,

    max_depth=5,

    random_state=42
)

model.fit(X, y_encoded)

print("✅ Tournament model trained.")
```

## 🏟️ Tournament Structure Analysis

Analyzing group-stage and knockout-stage tournament behavior.

```python
# ============================================
# 💪 Team Strength Table
# ============================================

team_strength = (

    match_df.groupby("home_team")[

        [
            "home_avg_points_last_5",
            "home_avg_goals_scored_last_5",
            "home_avg_goals_conceded_last_5",
            "home_avg_goal_diff_last_5",
            "home_win_streak",
            "home_clean_sheets_last_5",
            "home_momentum_score",
            "home_elo"
        ]

    ]

    .mean()
)

team_strength.head()
```

## 🤖 Match Prediction Analysis

Generating and evaluating football match predictions.

```python
# ============================================
# ⚽ Match Prediction Function
# ============================================

def predict_match(home_team, away_team):

    home = team_strength.loc[home_team]
    away = team_strength.loc[away_team]

    input_data = pd.DataFrame({

        "home_avg_points_last_5":
            [home["home_avg_points_last_5"]],

        "away_avg_points_last_5":
            [away["home_avg_points_last_5"]],

        "home_avg_goals_scored_last_5":
            [home["home_avg_goals_scored_last_5"]],

        "away_avg_goals_scored_last_5":
            [away["home_avg_goals_scored_last_5"]],

        "home_avg_goals_conceded_last_5":
            [home["home_avg_goals_conceded_last_5"]],

        "away_avg_goals_conceded_last_5":
            [away["home_avg_goals_conceded_last_5"]],

        "home_avg_goal_diff_last_5":
            [home["home_avg_goal_diff_last_5"]],

        "away_avg_goal_diff_last_5":
            [away["home_avg_goal_diff_last_5"]],

        "home_win_streak":
            [home["home_win_streak"]],

        "away_win_streak":
            [away["home_win_streak"]],

        "home_clean_sheets_last_5":
            [home["home_clean_sheets_last_5"]],

        "away_clean_sheets_last_5":
            [away["home_clean_sheets_last_5"]],

        "home_momentum_score":
            [home["home_momentum_score"]],

        "away_momentum_score":
            [away["home_momentum_score"]],

        "home_elo":
            [home["home_elo"]],

        "away_elo":
            [away["home_elo"]],

        "elo_difference":
            [
                home["home_elo"] -
                away["home_elo"]
            ]
    })

    prediction = model.predict(input_data)[0]

    probabilities = model.predict_proba(input_data)[0]

    result = encoder.inverse_transform(
        [prediction]
    )[0]

    probability_dict = dict(
        zip(
            encoder.classes_,
            probabilities
        )
    )

    return {

        "prediction": result,

        "home_win_probability":
            probability_dict.get(
                "Home Win", 0
            ),

        "draw_probability":
            probability_dict.get(
                "Draw", 0
            ),

        "away_win_probability":
            probability_dict.get(
                "Away Win", 0
            )
    }
```

```python
# ============================================
# 🌍 FIFA World Cup 2026 Projected Groups
# ============================================

groups = {

    "Group A": [
        "Mexico",
        "South Africa",
        "South Korea",
        "Czech Republic"
    ],

    "Group B": [
        "Canada",
        "Bosnia and Herzegovina",
        "Qatar",
        "Switzerland"
    ],

    "Group C": [
        "Brazil",
        "Morocco",
        "Haiti",
        "Scotland"
    ],

    "Group D": [
        "United States",
        "Paraguay",
        "Australia",
        "Turkey"
    ],

    "Group E": [
        "Germany",
        "Curaçao",
        "Ivory Coast",
        "Ecuador"
    ],

    "Group F": [
        "Netherlands",
        "Japan",
        "Sweden",
        "Tunisia"
    ],

    "Group G": [
        "Belgium",
        "Egypt",
        "Iran",
        "New Zealand"
    ],

    "Group H": [
        "Spain",
        "Cape Verde",
        "Saudi Arabia",
        "Uruguay"
    ],

    "Group I": [
        "France",
        "Senegal",
        "Bolivia",
        "Norway"
    ],

    "Group J": [
        "Argentina",
        "Algeria",
        "Austria",
        "Jordan"
    ],

    "Group K": [
        "Portugal",
        "DR Congo",
        "Uzbekistan",
        "Colombia"
    ],

    "Group L": [
        "England",
        "Croatia",
        "Ghana",
        "Panama"
    ]
}
```

## 🎮 Tournament Simulation

Simulating FIFA World Cup tournament scenarios and outcomes.

```python
# ============================================
# 🎲 Simulate Match
# ============================================

def simulate_match(home_team, away_team):

    prediction = predict_match(
        home_team,
        away_team
    )

    probabilities = [

        prediction["home_win_probability"],

        prediction["draw_probability"],

        prediction["away_win_probability"]
    ]

    outcomes = [

        "Home Win",
        "Draw",
        "Away Win"
    ]

    result = random.choices(

        outcomes,

        weights=probabilities,

        k=1
    )[0]

    return result
```

```python
# ============================================
# 🏆 Simulate Group Stage
# ============================================

group_tables = {}

for group_name, teams in groups.items():

    table = pd.DataFrame({

        "Team": teams,

        "Points": 0,

        "Wins": 0,

        "Draws": 0,

        "Losses": 0
    })

    for i in range(len(teams)):

        for j in range(i + 1, len(teams)):

            home = teams[i]
            away = teams[j]

            result = simulate_match(
                home,
                away
            )

            if result == "Home Win":

                table.loc[
                    table["Team"] == home,
                    "Points"
                ] += 3

                table.loc[
                    table["Team"] == home,
                    "Wins"
                ] += 1

                table.loc[
                    table["Team"] == away,
                    "Losses"
                ] += 1

            elif result == "Away Win":

                table.loc[
                    table["Team"] == away,
                    "Points"
                ] += 3

                table.loc[
                    table["Team"] == away,
                    "Wins"
                ] += 1

                table.loc[
                    table["Team"] == home,
                    "Losses"
                ] += 1

            else:

                table.loc[
                    table["Team"] == home,
                    "Points"
                ] += 1

                table.loc[
                    table["Team"] == away,
                    "Points"
                ] += 1

                table.loc[
                    table["Team"] == home,
                    "Draws"
                ] += 1

                table.loc[
                    table["Team"] == away,
                    "Draws"
                ] += 1

    table = table.sort_values(

        by=["Points", "Wins"],

        ascending=False
    ).reset_index(drop=True)

    group_tables[group_name] = table
```

```python
# ============================================
# 📊 Group Stage Results
# ============================================

for group_name, table in group_tables.items():

    print("\n")
    print("=" * 50)

    print(group_name)

    print("=" * 50)

    display(table)
```

```python
# ============================================
# ✅ Qualified Teams
# ============================================

qualified = []

for group_name, table in group_tables.items():

    qualified.extend(

        table["Team"]
        .head(2)
        .tolist()
    )

qualified
```

```python
# ============================================
# ⚔️ Quarterfinals
# ============================================

quarterfinals = [

    (qualified[0], qualified[3]),

    (qualified[1], qualified[2]),

    (qualified[4], qualified[7]),

    (qualified[5], qualified[6])
]

quarterfinal_winners = []

for home, away in quarterfinals:

    result = simulate_match(
        home,
        away
    )

    if result == "Home Win":

        winner = home

    elif result == "Away Win":

        winner = away

    else:

        winner = random.choice([
            home,
            away
        ])

    quarterfinal_winners.append(
        winner
    )

    print(
        f"{home} vs {away} → Winner: {winner}"
    )
```

```python
# ============================================
# 🏅 Semifinals
# ============================================

semifinals = [

    (
        quarterfinal_winners[0],
        quarterfinal_winners[1]
    ),

    (
        quarterfinal_winners[2],
        quarterfinal_winners[3]
    )
]

finalists = []

for home, away in semifinals:

    result = simulate_match(
        home,
        away
    )

    if result == "Home Win":

        winner = home

    elif result == "Away Win":

        winner = away

    else:

        winner = random.choice([
            home,
            away
        ])

    finalists.append(winner)

    print(
        f"{home} vs {away} → Finalist: {winner}"
    )
```

```python
# ============================================
# 🏆 FIFA WORLD CUP FINAL
# ============================================

home = finalists[0]
away = finalists[1]

result = simulate_match(
    home,
    away
)

if result == "Home Win":

    champion = home

elif result == "Away Win":

    champion = away

else:

    champion = random.choice([
        home,
        away
    ])

print("\n")
print("=" * 60)

print(
    f"🏆 WORLD CUP CHAMPION: {champion}"
)

print("=" * 60)
```
