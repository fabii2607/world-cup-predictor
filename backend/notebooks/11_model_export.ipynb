## ⚙️ Feature & Preprocessing Pipeline

Organizing preprocessing and feature transformation assets.

```python
# ============================================
# 📚 Import Libraries
# ============================================

import pandas as pd
import numpy as np

import joblib

from sklearn.ensemble import GradientBoostingClassifier
from sklearn.preprocessing import LabelEncoder

import warnings
warnings.filterwarnings("ignore")
```

## 🗂️ Dataset Preparation

Loading datasets and organizing export dependencies.

```python
# ============================================
# 📥 Load Enhanced Dataset
# ============================================

match_df = pd.read_csv(
    "../data/processed/match_dataset_enhanced.csv"
)

match_df.head()
```

```python
# ============================================
# ⚙️ Feature Selection
# ============================================

features = [

    "home_avg_points_last_5",
    "away_avg_points_last_5",

    "home_avg_goals_scored_last_5",
    "away_avg_goals_scored_last_5",

    "home_avg_goals_conceded_last_5",
    "away_avg_goals_conceded_last_5",

    "home_avg_goal_diff_last_5",
    "away_avg_goal_diff_last_5",

    "home_win_streak",
    "away_win_streak",

    "home_clean_sheets_last_5",
    "away_clean_sheets_last_5",

    "home_momentum_score",
    "away_momentum_score",

    "home_elo",
    "away_elo",

    "elo_difference"
]

X = match_df[features].fillna(0)

y = match_df["target"]
```

```python
# ============================================
# 🔤 Encode Target
# ============================================

encoder = LabelEncoder()

y_encoded = encoder.fit_transform(y)
```

## 🤖 Model Loading

Loading trained machine learning models and related assets.

```python
# ============================================
# 🚀 Train Final Model
# ============================================

model = GradientBoostingClassifier(

    n_estimators=200,

    learning_rate=0.05,

    max_depth=5,

    random_state=42
)

model.fit(X, y_encoded)

print("✅ Final model trained successfully.")
```

```python
# ============================================
# 📂 Create Models Folder
# ============================================

import os

os.makedirs(
    "../models",
    exist_ok=True
)
```

## 📦 Model Export Pipeline

Saving models and analytical assets for deployment workflows.

```python
# ============================================
# 💾 Export Trained Model
# ============================================

joblib.dump(

    model,

    "../models/world_cup_model.pkl"
)

print(
    "✅ Model exported successfully."
)
```

```python
# ============================================
# 💾 Export Label Encoder
# ============================================

joblib.dump(

    encoder,

    "../models/label_encoder.pkl"
)

print(
    "✅ Encoder exported successfully."
)
```

```python
# ============================================
# 💪 Team Strength Table
# ============================================

team_strength = (

    match_df.groupby("home_team")[

        [
            "home_avg_points_last_5",
            "home_avg_goals_scored_last_5",
            "home_avg_goals_conceded_last_5",
            "home_avg_goal_diff_last_5",
            "home_win_streak",
            "home_clean_sheets_last_5",
            "home_momentum_score",
            "home_elo"
        ]

    ]

    .mean()
)

team_strength.head()
```

```python
# ============================================
# 💾 Export Team Strength Table
# ============================================

team_strength.to_csv(

    "../data/processed/team_strength.csv"
)

print(
    "✅ Team strength table exported."
)
```

```python
# ============================================
# 💾 Export Feature List
# ============================================

joblib.dump(

    features,

    "../models/model_features.pkl"
)

print(
    "✅ Feature list exported."
)
```
